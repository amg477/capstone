```python
# 4) Attribution including keywords for sourcetype 
keywords = pd.read_csv("../data/processed/top_keywords_300_list.csv", header=None)[0].dropna().tolist()

pub_kw_attr = run_markov_attribution(
    df=df,
    path_key="tag_name",
    time_col="load_date",
    state_col="source_type_name",
    weight_col="vipr_weight",
    keyword_mode=True,
    keywords=keywords,   
)
```

```python
# Collect results in a list
results = []

# Add each attribution run (make sure the variables exist in your notebook)
pub_attr["dimension"] = "publication_name"
results.append(pub_attr)

pubs_attr["dimension"] = "publisher_name"
results.append(pubs_attr)

stype_attr["dimension"] = "source_type_name"
results.append(stype_attr)

author_attr["dimension"] = "author_name"
results.append(author_attr)

channel_attr["dimension"] = "channel_name"
results.append(channel_attr)

feed_attr["dimension"] = "feed_name"
results.append(feed_attr)

genre_attr["dimension"] = "genre"
results.append(genre_attr)

# Keyword-based attribution (already run as pub_kw_attr)
pub_kw_attr["dimension"] = "source_type_name"
pub_kw_attr["keyword_mode"] = True

# If it contains publication::keyword, split it
if "::" in pub_kw_attr["state"].iloc[0]:
    pub_kw_attr[["state_value","keyword"]] = pub_kw_attr["state"].str.split("::", n=1, expand=True)

results.append(pub_kw_attr)
```

```python
# Combine all into one tidy DataFrame
all_attr = pd.concat(results, axis=0, ignore_index=True)
```

```python
# Save a single CSV
all_attr.to_csv("../data/processed/attribution_all_dimensions.csv", index=False)
all_attr.head(20)
```
